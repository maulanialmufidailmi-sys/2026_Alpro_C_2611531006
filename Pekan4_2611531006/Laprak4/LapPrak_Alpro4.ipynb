## 2.1 Dasar Teori
# Pengertian Conditional Statement

Conditional statement adalah struktur dalam pemrograman yang digunakan untuk mengeksekusi blok kode tertentu berdasarkan suatu kondisi, sehingga program dapat menjalankan aksi yang berbeda-beda sesuai situasi yang dihadapi. Dalam Python, struktur ini diwujudkan melalui tiga pernyataan utama:

- **`if`** – mengeksekusi kode jika kondisi bernilai benar (True).
- **`elif`** – memeriksa kondisi tambahan jika kondisi sebelumnya salah.
- **`else`** – mengeksekusi kode jika semua kondisi sebelumnya salah.
## 2.2 Kode Program

### 2.2.1 Program if1_2611531006
</div>

```python
# Buat file dengan nama if1_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_1006 = float(input("Input IPK Anda = "))

if ipk_1006 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK",str(ipk_1006))

print("Program Selesai")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
Program `if1_1006.py` — contoh struktur kondisional `if` sederhana:

1. **Input**: `input()` meminta pengguna mengetik IPK, lalu `float()` mengubahnya jadi angka desimal, disimpan di `ipk_1006`.
2. **Kondisi if**: mengecek apakah `ipk_1006 > 2.75`. Kalau **benar**, baris `print` di bawahnya (yang terindentasi) dijalankan, menampilkan pesan "Anda Lulus Sangat Memuaskan" beserta nilai IPK-nya.
3. **Kalau kondisi salah** (IPK ≤ 2.75), baris pesan itu **dilewati**, program langsung lanjut tanpa menampilkan apa-apa terkait kelulusan.
4. **Baris terakhir** `print("Program Selesai")` berada di luar blok `if` (tidak terindentasi), jadi selalu dijalankan di akhir, apa pun hasil pengecekan IPK.

Intinya: ini contoh **if satu cabang** (tanpa `elif`/`else`) — hanya ada satu jalur pesan khusus, sisanya langsung ke penutup program.


### 2.2.2 Program if2_2611531006
</div>

```python
# Buat file dengan nama if2_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_1006 = float(input("Input IPK Anda = "))

if ipk_1006 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK",str(ipk_1006))
else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
Program `if2_1006.py` — contoh struktur kondisional **if-else**:

1. **Input**: `input()` meminta pengguna mengetik IPK, lalu `float()` mengubahnya jadi angka desimal (bukan string), disimpan ke variabel `ipk_1006`.
2. **Kondisi if**: mengecek apakah `ipk_1006 > 2.75`. Kalau **benar**, baris di bawahnya (terindentasi) dijalankan → mencetak "Anda Lulus Sangat Memuaskan dengan IPK" beserta nilainya.
3. **Else**: ini bagian baru dibanding `if1` — kalau kondisi `if` **salah** (IPK ≤ 2.75), program otomatis masuk ke blok `else` dan mencetak "Anda Tidak Lulus".
4. **Sifat if-else**: hanya **satu** dari dua blok (`if` atau `else`) yang pernah dijalankan dalam satu kali proses program, tidak mungkin keduanya jalan sekaligus, dan tidak mungkin keduanya dilewati.
5. **Baris terakhir** `print("Program Selesai")` berada di luar blok if-else (tidak terindentasi), jadi selalu dijalankan di akhir, apa pun hasil pengecekan IPK.

**Perbedaan dengan `if1`:**
- `if1` hanya punya satu jalur pesan (kalau kondisi salah, tidak ada pesan sama sekali).
- `if2` memastikan **setiap kemungkinan nilai IPK punya pesan yang sesuai** — baik lulus maupun tidak lulus, tidak ada kasus yang "diam saja".

### 2.2.3 Program multi_if1_2611531006
</div>

```python
# Buat file dengan nama multi_if1_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_1006 = int(input("Input Umur Anda = "))
sim_1006 = input("Apakah Anda Sudah Punya SIM (y/t): ")[0]

if umur_1006 >= 17 and sim_1006 == "y":
    print("Anda Sudah dewasa dan boleh bawa motor")

if umur_1006 >= 17 and sim_1006 != "y":
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")

if umur_1006 < 17 and sim_1006 != "y":
    print("Anda Belum Cukup Umur bawa motor")

if umur_1006 < 17 and sim_1006 == "y":
    print("Anda Belum Cukup Umur punya SIM")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. **Input umur**: `int(input(...))` meminta umur, diubah jadi bilangan bulat, disimpan di `umur_1006`.
2. **Input SIM**: `input(...)[0]` meminta jawaban y/t, lalu `[0]` mengambil **karakter pertama** saja dari jawaban tersebut (misalnya kalau user mengetik "ya", yang diambil hanya `"y"`), disimpan di `sim_1006`.
3. **Empat blok `if` terpisah** (bukan `if-elif-else`), masing-masing dicek satu per satu secara independen, menggabungkan dua syarat sekaligus dengan `and`:
   - `umur >= 17 and sim == "y"` → sudah dewasa **dan** punya SIM → boleh bawa motor.
   - `umur >= 17 and sim != "y"` → sudah dewasa **tapi** belum punya SIM → belum boleh bawa motor.
   - `umur < 17 and sim != "y"` → belum cukup umur **dan** belum punya SIM → belum boleh bawa motor.
   - `umur < 17 and sim == "y"` → belum cukup umur tapi sudah punya SIM (kasus ganjil) → belum cukup umur untuk punya SIM.
4. **Operator `and`**: kondisi hanya bernilai True jika **kedua syarat** di kiri dan kanan `and` sama-sama terpenuhi.
5. Karena keempat blok memakai `if` biasa (bukan `elif`), Python **mengecek semuanya satu per satu** — meski begitu, karena keempat kondisi ini saling eksklusif (tidak mungkin dua-duanya benar sekaligus), hanya **satu pesan** yang akan tercetak setiap kali program dijalankan.

### 2.2.4 Program multi_if2_2611531006
</div>

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dan user
total_belanja_1006 = float(input("Input Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y atau 't')
input_member_1006 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_1006 = input_member_1006 in ["y", "t"]

# Input satus kode promo (mengecek apakah user mengetik 'y atau 't')
input_promo_1006 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1006 = input_promo_1006 in ["y", "t"]

total_diskon_persen_1006 = 0

# Multi-if terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_1006 > 1000000:
    total_diskon_persen_1006 += 10  # Diskon total besar

if is_member_1006:
    total_diskon_persen_1006 += 5  # Diskon member

if kode_promo_valid_1006:
    total_diskon_persen_1006 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_1006 = total_belanja_1006 * (total_diskon_persen_1006 / 100)
total_bayar_1006 = total_belanja_1006 - nominal_diskon_1006

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_1006} (Rp {nominal_diskon_1006:.0f})")
print(f"Total Bayar   : Rp {total_bayar_1006:.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_1006}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

1. **Input total belanja**: `float(input(...))` menyimpan nominal belanja ke `total_belanja_1006`.
2. **Input status member**: `.strip().lower()` membersihkan spasi dan mengubah ke huruf kecil, lalu dicek dengan `in ["y", "t"]` → hasilnya `True`/`False` disimpan di `is_member_1006`.
3. **Input kode promo**: caranya sama, hasilnya `True`/`False` disimpan di `kode_promo_valid_1006`.
4. **Variabel akumulator** `total_diskon_persen_1006 = 0` — penampung diskon, akan bertambah sesuai syarat yang terpenuhi.
5. **Tiga blok `if` terpisah** (bukan `elif`), masing-masing dicek independen dan diskonnya **ditambahkan (`+=`)** jika syarat terpenuhi:
   - Belanja > Rp1.000.000 → +10%
   - Member → +5%
   - Kode promo valid → +15%
6. **Hitung hasil**: nominal diskon = belanja × (persen/100), total bayar = belanja − nominal diskon.
7. **Output** ditampilkan pakai f-string dengan `.0f` untuk membulatkan nominal ke angka bulat.

Perbedaan dengan multi_if1_1006.py

- **Sifat kondisi**: di `multi_if1`, keempat kondisi saling eksklusif (tidak mungkin dua-duanya benar sekaligus); di `multi_if2`, ketiga kondisi bersifat independen — bisa lebih dari satu benar dalam satu kali run.
- **Efek `if` terpisah**: di `multi_if1`, karena kondisinya eksklusif, hasilnya berperilaku seperti `elif` — hanya satu pesan yang tercetak; di `multi_if2`, karena kondisinya bisa tumpang tindih, efeknya jadi **akumulasi/penumpukan** nilai diskon.
- **Variabel hasil**: `multi_if1` tidak punya penampung, langsung `print` di tiap blok `if`; `multi_if2` punya variabel akumulator (`total_diskon_persen_1006`) yang nilainya terus bertambah tiap syarat lolos.
- **Operator kondisi**: `multi_if1` memakai `and` untuk menggabungkan dua syarat sekaligus dalam satu `if`; `multi_if2` tiap `if` hanya mengecek satu syarat, tanpa `and`.
- **Kasus penggunaan**: `multi_if1` cocok untuk menentukan satu kategori/status dari beberapa kemungkinan; `multi_if2` cocok untuk menghitung total/skor dari beberapa syarat yang bisa terpenuhi bersamaan.

### 2.2.5 Program if_elif_else1_2611531006

4. **`elif` pertama**: cek `umur >= 17 and sim != "y"` → dewasa tapi belum punya SIM.
5. **`elif` kedua**: cek `umur < 17 and sim == "y"` → belum cukup umur tapi sudah punya SIM.
6. **`else`**: menangkap **semua kemungkinan sisanya** (otomatis berarti `umur < 17 and sim != "y"`) → belum cukup umur dan tidak boleh bawa motor.
7. **`print("Program Selesai")`**: di luar blok, selalu dijalankan di akhir.

### 2.2.6 Program nilai_2611531006

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai_1006 = int(input("Input nilai angka: "))

if nilai_1006 >= 81:
    print("A")
elif nilai_1006 >= 70:
    print("B")
elif nilai_1006 >= 60:
    print("C")
elif nilai_1006 >= 50:
    print("D")
else:
    print("E")
```

4. **`elif` kedua**: cek `nilai >= 60` → nilai antara 60–69, cetak "C".
5. **`elif` ketiga**: cek `nilai >= 50` → nilai antara 50–59, cetak "D".
6. **`else`**: menangkap sisanya (nilai < 50), cetak "E".

### 2.2.7 Program match_case1_2611531006
</div>

```python
# Buat file dengan nama case1.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_1006 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_1006:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

1. **Input bulan**: `int(input(...))` meminta pengguna mengetik angka bulan, diubah jadi bilangan bulat dengan `int()`, disimpan ke variabel `bulan_1006`.
2. **`match bulan_1006:`** — membuka struktur pencocokan (pattern matching), nilai yang tersimpan di `bulan_1006` akan dibandingkan dengan tiap `case` di bawahnya secara berurutan dari atas ke bawah.
3. **`case 1:`** — jika `bulan_1006` bernilai `1`, maka baris di bawahnya (`print("Januari")`) dijalankan.
4. **`case 2:` sampai `case 12:`** — pola yang sama berulang untuk setiap angka bulan, masing-masing langsung menampilkan nama bulan yang sesuai (Februari, Maret, April, dan seterusnya).
5. **Sifat "berhenti setelah cocok"**: begitu Python menemukan satu `case` yang cocok dengan nilai `bulan_1006`, kode di dalamnya dijalankan dan Python **langsung keluar** dari struktur `match`, tidak lagi memeriksa `case` yang tersisa di bawahnya.
6. **`case _:`** — tanda garis bawah `_` adalah simbol **wildcard**, artinya "apa saja yang belum cocok di atas". Bagian ini berfungsi sebagai penangkap terakhir/default.
7. **Kapan `case _` dijalankan**: hanya jika `bulan_1006` bernilai di luar 1–12, misalnya `0`, `13`, `-5`, atau angka besar lainnya → program mencetak "Angka tidak valid".
8. **Tidak perlu `break`**: berbeda dengan bahasa lain seperti C/Java yang butuh `break` di setiap case agar tidak "jatuh" ke case berikutnya, di Python struktur `match-case` **otomatis berhenti** setelah satu case cocok, tanpa perlu ditulis eksplisit.
9. **Ketersediaan versi**: fitur `match-case` ini baru tersedia mulai **Python 3.10** ke atas, jadi tidak bisa dipakai di versi Python yang lebih lama.

```python
# =========================================================
# SISTEM LOKET TERPADU & AUDIT TRANSAKSI EKSPEDISI WAHANA
# =========================================================

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# -------------------------------
# 1. Input Data Pengunjung
# -------------------------------
nama_1006 = input("Masukkan Nama Pengunjung        : ")
umur_1006 = int(input("Input umur anda                 : "))
sim_1006 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].strip().lower()
jumlah_tiket_1006 = int(input("Masukkan jumlah tiket           : "))

# if tunggal - validasi kelogisan jumlah tiket
if jumlah_tiket_1006 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

# -------------------------------
# 2. Pemilihan Wahana (match-case)
# -------------------------------
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_1006 = int(input("Masukkan nomor paket (1-5)      : "))

match paket_1006:
    case 1:
        nama_wahana_1006 = "Wahana Safari Rimba"
        harga_satuan_1006 = 50000
    case 2:
        nama_wahana_1006 = "Wahana Arung Jeram"
        harga_satuan_1006 = 75000
    case 3:
        nama_wahana_1006 = "Wahana Motor ATV Ekstrim"
        harga_satuan_1006 = 120000
    case 4:
        nama_wahana_1006 = "Wahana Roller Coaster Kilat"
        harga_satuan_1006 = 100000
    case 5:
        nama_wahana_1006 = "Wahana All-Access VIP"
        harga_satuan_1006 = 220000
    case _:
        print("Paket wahana tidak valid!")
        sys.exit()

# -------------------------------
# 3. Input Member & Promo
# -------------------------------
is_member_1006 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_1006 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

# -------------------------------
# 4. Validasi Izin Kendali Wahana
#    (if - elif - else, operator logika and/!=)
# -------------------------------
if paket_1006 == 3 and umur_1006 >= 17 and sim_1006 == 'y':
    status_akses_1006 = "Anda sudah dewasa dan boleh mengendarai ATV sendiri."
elif paket_1006 == 3 and umur_1006 >= 17 and sim_1006 != 'y':
    status_akses_1006 = "Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur)."
elif paket_1006 == 3 and umur_1006 < 17 and sim_1006 == 'y':
    status_akses_1006 = "Identitas tidak valid: Belum cukup umur memiliki SIM."
elif paket_1006 == 3 and umur_1006 < 17 and sim_1006 != 'y':
    status_akses_1006 = "Anda belum cukup umur dan tidak boleh bawa motor ATV."
elif paket_1006 != 3 and umur_1006 >= 10:
    status_akses_1006 = "Anda memenuhi syarat umur untuk menikmati wahana ini."
else:
    status_akses_1006 = "Anda belum cukup umur untuk menikmati wahana ini."

# -------------------------------
# 5. Akumulasi Diskon Bertingkat
#    (Multi-IF terpisah, bisa ditumpuk)
# -------------------------------
subtotal_1006 = harga_satuan_1006 * jumlah_tiket_1006
total_diskon_persen_1006 = 0

if subtotal_1006 >= 200000:
    total_diskon_persen_1006 += 10  # Diskon Belanja Besar

if is_member_1006 in ['y', 'ya']:
    total_diskon_persen_1006 += 5  # Diskon Member

if kode_promo_valid_1006 in ['y', 'ya']:
    total_diskon_persen_1006 += 15  # Diskon Voucher Promo

if jumlah_tiket_1006 >= 5:
    total_diskon_persen_1006 += 5  # Diskon Tambahan Rombongan

nominal_diskon_1006 = subtotal_1006 * (total_diskon_persen_1006 / 100)
total_bayar_1006 = subtotal_1006 - nominal_diskon_1006

# -------------------------------
# 6. Evaluasi Kelulusan Audit (if - else)
# -------------------------------
if total_bayar_1006 > 300000:
    catatan_layanan_1006 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_1006 = "Terima kasih telah berkunjung."

# -------------------------------
# 7. Output Rincian
# -------------------------------
print(f"\n--- KELAYAKAN PENGENDARA WAHANA ---")
print(f"Status Akses: {status_akses_1006}")

print(f"\n--- Rincian Pembayaran ---")
print(f"Wahana Dipilih   : {nama_wahana_1006}")
print(f"Subtotal Belanja : Rp {subtotal_1006:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_1006}% (Rp {nominal_diskon_1006:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_1006:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_1006}")
print("Program Selesai")
```
