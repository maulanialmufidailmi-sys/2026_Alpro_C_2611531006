## 2.1 Dasar Teori
# Pengertian Conditional Statement

Conditional statement adalah struktur dalam pemrograman yang digunakan untuk mengeksekusi blok kode tertentu berdasarkan suatu kondisi, sehingga program dapat menjalankan aksi yang berbeda-beda sesuai situasi yang dihadapi. Dalam Python, struktur ini diwujudkan melalui tiga pernyataan utama:

- **`if`** – mengeksekusi kode jika kondisi bernilai benar (True).
- **`elif`** – memeriksa kondisi tambahan jika kondisi sebelumnya salah.
- **`else`** – mengeksekusi kode jika semua kondisi sebelumnya salah.
## 2.2 Kode Program

### 2.2.1 Program if1_2611531006
</div>

```python
# Buat file dengan nama if1_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_1021 = float(input("Input IPK Anda = "))

if ipk_1021 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK",str(ipk_1021))

print("Program Selesai")
```

### 2.2.4 Program multi_if2_2611531021
</div>

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dan user
total_belanja_1021 = float(input("Input Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y atau 't')
input_member_1021 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_1021 = input_member_1021 in ["y", "t"]

# Input satus kode promo (mengecek apakah user mengetik 'y atau 't')
input_promo_1021 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1021 = input_promo_1021 in ["y", "t"]

total_diskon_persen_1021 = 0

# Multi-if terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_1021 > 1000000:
    total_diskon_persen_1021 += 10  # Diskon total besar

if is_member_1021:
    total_diskon_persen_1021 += 5  # Diskon member

if kode_promo_valid_1021:
    total_diskon_persen_1021 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_1021 = total_belanja_1021 * (total_diskon_persen_1021 / 100)
total_bayar_1021 = total_belanja_1021 - nominal_diskon_1021

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_1021} (Rp {nominal_diskon_1021:.0f})")
print(f"Total Bayar   : Rp {total_bayar_1021:.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_1021}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Program dimulai dengan meminta pengguna memasukkan **total belanja** dalam rupiah menggunakan fungsi `input()`.
2. Nilai total belanja diubah menjadi tipe data `float` dan disimpan dalam variabel `total_belanja_1021`.
3. Program meminta pengguna memasukkan status member dengan pilihan `y` atau `t`. Input kemudian diubah menjadi huruf kecil menggunakan `.lower()` dan spasi yang tidak diperlukan dihapus menggunakan `.strip()`.
4. Program memeriksa apakah input member berupa `"y"` atau `"t"` menggunakan operator `in`. Hasil pemeriksaan disimpan dalam variabel `is_member_1021`.
5. Program meminta pengguna memasukkan status kode promo dengan pilihan `y` atau `t`. Input diproses dengan cara yang sama dan hasilnya disimpan dalam variabel `kode_promo_valid_1021`.
6. Variabel `total_diskon_persen_1021` diberikan nilai awal `0` sebagai tempat untuk menyimpan jumlah persentase diskon.
7. Program memeriksa kondisi pertama. Jika total belanja lebih dari **Rp1.000.000**, program menambahkan diskon sebesar **10%**.
8. Program memeriksa kondisi kedua. Jika pengguna merupakan member, program menambahkan diskon sebesar **5%**.
9. Program memeriksa kondisi ketiga. Jika kode promo dianggap valid, program menambahkan diskon sebesar **15%**.
10. Karena menggunakan beberapa `if` yang terpisah, setiap kondisi diperiksa secara independen. Jika beberapa kondisi terpenuhi, persentase diskon akan **diakumulasikan**.
11. Program menghitung nominal diskon dengan mengalikan total belanja dengan persentase diskon yang telah diperoleh.
12. Program menghitung total pembayaran dengan mengurangi total belanja dengan nominal diskon.
13. Program menampilkan rincian pembayaran berupa total persentase diskon, nominal diskon dalam rupiah, dan total yang harus dibayar.
14. Program menampilkan kembali jumlah diskon yang diperoleh dalam bentuk persentase.

### 2.2.7 Program match_case1_2611531021
</div>

```python
# Buat file dengan nama case1.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_1021 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_1021:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Program dimulai dengan meminta pengguna memasukkan angka bulan dari **1 sampai 12** menggunakan fungsi `input()`.
2. Nilai yang dimasukkan diubah menjadi tipe data `int` dan disimpan dalam variabel `bulan_1021`.
3. Program menggunakan struktur **`match-case`** untuk mencocokkan nilai `bulan_1021` dengan beberapa pilihan yang tersedia.
4. Jika nilai yang dimasukkan adalah **1**, program menampilkan **Januari**.
5. Jika nilai **2 sampai 12**, program menampilkan nama bulan yang sesuai, mulai dari Februari hingga Desember.
6. Jika angka yang dimasukkan tidak sesuai dengan angka **1 sampai 12**, program menjalankan `case _` dan menampilkan pesan **"Angka tidak valid"**.
7. Setelah nama bulan atau pesan kesalahan ditampilkan, program selesai.

```python
("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")
nama_1021 = input("Masukkan Nama Pengunjung        : ")
umur_1021 = int(input("Input umur anda                 : "))
sim_c_1021 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].strip().lower()

print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_1021 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_1021 = int(input("Masukkan jumlah tiket           : "))
if jumlah_tiket_1021 <= 0:
    print("Jumlah tiket tidak valid!")
    exit()
member_1021 = input("Apakah Anda member? (y/t)       : ")[0].strip().lower()
promo_1021 = input("Apakah kode promo valid? (y/t)  : ")[0].strip().lower()

wahana_1021 = ""
harga_satuan_1021 = 0

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
match paket_1021:
    case 1: # Wahana Safari Rimba
        if umur_1021 >= 10:
            print("Anda sudah dewasa dan boleh mengendarai Wahana Safari Rimba sendiri.")
        else:
            print("Anda wajib didampingi orang tua.")
        harga_satuan_1021 = 50000
    case 2: # Wahana Arung Jeram
        if umur_1021 >= 15:
            print("Anda sudah dewasa dan boleh mengendarai Wahana Arung Jeram sendiri.")
        else:
            print("Anda wajib didampingi orang tua.")
        harga_satuan_1021 = 75000
    case 3: # Wahana Motor ATV Ekstrim
        if umur_1021 >= 17 and sim_c_1021 == 'y':
            print("Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
        elif umur_1021 >= 17 and sim_c_1021 != 'y':
            print("Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
        elif umur_1021 < 17 and sim_c_1021 == 'y':
            print("Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Anda belum cukup umur dan tidak boleh bawa motor ATV.")
        harga_satuan_1021 = 120000
    case 4: # Wahana Roller Coaster Kilat
        if umur_1021 >= 17:
            print("Anda sudah dewasa dan boleh mengendarai Wahana Roller Coaster Kilat sendiri.")
        else:
            print("Anda wajib didampingi orang tua.")
        harga_satuan_1021 = 100000
    case 5: # Wahana All-Access VIP
        if umur_1021 >= 17 and sim_c_1021 == 'y':
            print("Anda sudah dewasa dan boleh mengendarai semua wahana sendiri.")
        elif umur_1021 >= 17 and sim_c_1021 != 'y':
            print("Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
        elif umur_1021 < 17 and sim_c_1021 == 'y':
            print("Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Anda belum cukup umur dan tidak boleh bawa motor ATV.")
        harga_satuan_1021 = 220000
    case _: 
        print("Paket wahana tidak valid!")
        exit()

diskon_1021 = 0
if harga_satuan_1021 * jumlah_tiket_1021 >= 200000:
    diskon_1021 += 10
if member_1021 in ["y","ya"]:
    diskon_1021 += 5
if promo_1021 in ["y","ya"]:
    diskon_1021 += 15
if jumlah_tiket_1021 >= 5:
    diskon_1021 += 5

nominal_1021 = harga_satuan_1021 * jumlah_tiket_1021
nominal_diskon_1021 = nominal_1021 * (diskon_1021 / 100)
total_bayar_1021 = nominal_1021 - nominal_diskon_1021
print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {nominal_1021:,}")
if nominal_1021 > 300000: 
    print("Selamat! Anda berhak mendapatkan Souvenir Gratis.")
print(f"Total Diskon     : {diskon_1021}% (Rp {int(nominal_diskon_1021):,})")
print(f"Total Bayar      : Rp {int(total_bayar_1021):,}")
print("Catatan Layanan  : Terima kasih telah berkunjung.")
print("Program Selesai")
```
