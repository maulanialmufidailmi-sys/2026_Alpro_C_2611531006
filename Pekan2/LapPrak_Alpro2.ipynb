## 2.1 Dasar Teori

Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub menyajikan materi awal seputar cara kerja bahasa pemrograman Python, pengoperasian Visual Studio Code selaku aplikasi penyunting kode, serta pemanfaatan GitHub sebagai platform penyimpanan dan pengelolaan kode program secara daring.

Praktikum Tipe Data dan Variabel menyajikan materi awal pada bahasa pemrograman Python terkait cara data disimpan dan diolah di dalam program. Tipe data berguna untuk menentukan kategori nilai yang boleh diproses, sedangkan variabel berguna sebagai media penampung atau acuan terhadap suatu nilai tertentu.

1. Integer (int)<br>
Integer ialah kategori data yang dipergunakan untuk menampung angka bulat, seperti 10, 25, maupun -5.
2. Float (float)<br>
Float ialah kategori data yang dipergunakan untuk menampung angka pecahan atau desimal, seperti 3.14, 10.5, maupun 0.75.
3. Complex (complex)<br>
Complex ialah kategori data yang dipergunakan untuk menampung angka kompleks yang tersusun atas komponen real dan imajiner. Sebagai gambaran, 3+2j.
4. String (str)<br>
String ialah kategori data yang dipergunakan untuk menampung tulisan atau gabungan karakter. Cara penulisannya diapit tanda petik, seperti "Python" maupun "Informatika".
5. Boolean (bool)<br>
Boolean ialah kategori data yang cuma mempunyai dua pilihan nilai, yaitu True serta False. Kategori data ini kerap dipakai untuk menggambarkan suatu keadaan benar ataupun keliru.
6. Konstanta<br>
Konstanta ialah suatu angka atau nilai yang pada dasarnya tak boleh berubah selagi program berjalan. Python tak memiliki aturan penulisan tersendiri untuk konstanta, hanya saja penamaan variabel memakai huruf besar kerap dipakai sebagai ciri bahwa variabel itu diperlakukan layaknya konstanta, seperti PI = 3.14.
7. Variabel<br>
Variabel berperan sebagai media penampung atau acuan terhadap suatu nilai. Di Python, penciptaan variabel dilakukan lewat pemberian nilai memakai tanda =, seperti nama = "Maulani" ataupun umur = 18.
8. Input dan Output<br>
Python menyediakan fungsi input() guna menampung data yang diketik pengguna serta print() guna menayangkan data pada layar. Data keluaran input() secara bawaan berkategori str, sehingga mesti diubah kategorinya lebih dulu apabila data itu mau dipakai selaku angka.

### 2.2.5 Program Konstanta_2611531006
</div>

```python
# Buat file dengan nama Konstanta_2611531006
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1006

from typing import Final
PI: Final = 3.14
print("Pi: %f" % (PI))
jari_1006 = float(input("Masukkan nilai jari-jari: "))
luas_1006 = PI * jari_1006
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1006,luas_1006))
```

5. Nilai luas_1021 = PI * jari_1006
