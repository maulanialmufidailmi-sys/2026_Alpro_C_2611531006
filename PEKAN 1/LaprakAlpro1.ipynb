## 2.1 Dasar Teori
Berikut pengertian untuk setiap poin, disusun berdasarkan sumber yang telah dicantumkan:

**1. Python**
Python adalah bahasa pemrograman tingkat tinggi yang bersifat *interpretatif* (dieksekusi langsung tanpa proses kompilasi) dan mendukung berbagai paradigma pemrograman, termasuk pemrograman terstruktur, berorientasi objek, dan fungsional. Menurut laman resminya, Python adalah bahasa pemrograman tingkat tinggi yang diinterpretasikan, berorientasi objek, dan memiliki semantik dinamis, sehingga cocok digunakan untuk berbagai keperluan seperti pengembangan aplikasi, pengolahan data, hingga kecerdasan buatan.

**2. Visual Studio Code (VS Code)**
VS Code adalah source code editor yang ringan namun memiliki fitur pengembangan yang cukup lengkap untuk membangun dan melakukan debugging aplikasi web maupun cloud modern, tersedia secara gratis di Linux, macOS, dan Windows. VS Code mendukung berbagai bahasa pemrograman termasuk Python melalui ekstensi tambahan.

**3. Git**
Git adalah sistem version control terdistribusi yang bersifat gratis dan open source, dirancang untuk menangani proyek mulai dari yang kecil hingga sangat besar dengan cepat dan efisien. Git memungkinkan setiap pengguna menyimpan riwayat perubahan kode secara lokal di komputer masing-masing.

**4. GitHub**
GitHub adalah platform pengembang yang lengkap untuk membangun, mengembangkan, dan mengirimkan perangkat lunak, yang penggunaan utamanya adalah untuk pengembangan software dan version control menggunakan Git. GitHub memungkinkan proyek diakses dari berbagai perangkat serta mendukung kolaborasi antar pengembang.

**5. Repository**
Repository adalah tempat penyimpanan seluruh berkas, folder, serta riwayat perubahan dari suatu proyek yang dikelola menggunakan Git. Repository dapat bersifat lokal (tersimpan di komputer pengguna) maupun *remote* (tersimpan pada server daring seperti GitHub).

**6. Commit**
Commit adalah proses menyimpan perubahan yang telah dilakukan pada berkas proyek ke dalam riwayat repository Git secara permanen, disertai pesan singkat (*commit message*) yang menjelaskan perubahan tersebut.

**7. Push**
Push adalah proses mengunggah perubahan yang telah di-*commit* dari repository lokal ke repository *remote* (seperti GitHub), sehingga perubahan tersebut dapat diakses dan digunakan bersama oleh anggota tim lain.
