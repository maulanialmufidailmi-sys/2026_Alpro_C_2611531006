## 2.1 Dasar Teori
Dalam bahasa pemrograman Python, operator adalah simbol khusus yang digunakan untuk melakukan suatu operasi atau manipulasi terhadap satu, dua, atau lebih nilai (disebut operand). Operator merupakan salah satu komponen fundamental dalam pemrograman karena hampir setiap program membutuhkan proses perhitungan, pengambilan keputusan, maupun manipulasi data yang melibatkan penggunaan operator.

Secara umum, operator dalam Python berfungsi untuk menghasilkan suatu nilai baru berdasarkan operand yang diproses. Sebagai contoh, operator dapat digunakan untuk menjumlahkan dua angka, membandingkan dua nilai untuk menentukan mana yang lebih besar, atau menggabungkan beberapa kondisi logika untuk pengambilan keputusan dalam program.

Python menyediakan berbagai jenis operator yang dikelompokkan berdasarkan fungsinya, antara lain:

### 1. Operator Aritmatika
Operator aritmatika digunakan untuk melakukan operasi matematis dasar antara dua nilai atau lebih. Python menyediakan beberapa operator aritmatika, yaitu:
- `+` (penjumlahan)
- `-` (pengurangan)
- `*` (perkalian)
- `/` (pembagian, menghasilkan bilangan desimal/float)
- `//` (pembagian bulat/floor division, hasil dibulatkan ke bawah)
- `%` (modulus, menghasilkan sisa pembagian)
- `**` (pemangkatan/eksponen)

### 2. Operator Assignment (Penugasan)
Operator penugasan digunakan untuk memberikan nilai ke suatu variabel. Selain operator `=` yang sederhana, Python juga menyediakan *compound assignment* yang menggabungkan operasi aritmatika dengan penugasan, seperti `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, dan `**=`. Operator ini berguna untuk memperbarui nilai variabel secara ringkas tanpa menulis ulang nama variabelnya.

### 3. Operator Bitwise
Operator bitwise bekerja pada representasi biner (bit) dari suatu bilangan bulat. Operator ini meliputi:
- `&` (AND) — menghasilkan 1 jika kedua bit bernilai 1
- `|` (OR) — menghasilkan 1 jika salah satu bit bernilai 1
- `^` (XOR) — menghasilkan 1 jika kedua bit berbeda
- `~` (NOT) — membalik seluruh bit (komplemen)
- `<<` (shift left) — menggeser bit ke kiri
- `>>` (shift right) — menggeser bit ke kanan

Operator ini sering digunakan dalam pemrograman tingkat rendah, manipulasi flag, dan optimasi.

### 4. Operator Logika (Logical Operator)
Operator logika digunakan untuk menggabungkan atau mengevaluasi ekspresi kondisional yang menghasilkan nilai boolean (`True`/`False`). Terdiri dari:
- `and` — bernilai True jika kedua kondisi benar
- `or` — bernilai True jika salah satu kondisi benar
- `not` — membalik nilai boolean

Operator ini banyak digunakan dalam struktur kontrol seperti percabangan (`if`) dan perulangan.

### 5. Operator Perbandingan (Comparison Operator)
Operator perbandingan digunakan untuk membandingkan dua nilai, dan hasilnya selalu berupa nilai boolean. Operatornya meliputi:
- `==` (sama dengan)
- `!=` (tidak sama dengan)
- `>` (lebih besar dari)
- `<` (lebih kecil dari)
- `>=` (lebih besar sama dengan)
- `<=` (lebih kecil sama dengan)

### 6. Operator Lainnya
Kategori ini biasanya mencakup operator khusus Python seperti:
- **Operator Identitas** (`is`, `is not`) — membandingkan apakah dua variabel merujuk ke objek yang sama di memori
- **Operator Keanggotaan** (`in`, `not in`) — memeriksa apakah suatu nilai terdapat dalam suatu urutan (list, string, tuple, dll.)

### 2.2.1 Program aritmatika_2611531006

````python
# Buat program untuk operator aritmatika dalam Python

angka1_1006 = int(input("Input angka-1: "))
angka2_1006 = int(input("Input angka-2: "))

# Penjumlahan
hasil_1006 = angka1_1006 + angka2_1006
print("\nOperator Penjumlahan")
print("Hasil =",hasil_1006)

# Pengurangan
hasil_1006 = angka1_1006 - angka2_1006
print("\nOperator Pengurangan")
print("Hasil =",hasil_1006)

# Perkalian
hasil_1006 = angka1_1006 * angka2_1006
print("\nOperator Perkalian")
print("Hasil =",hasil_1006)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_1006 != 0:
    hasil_1006 = angka1_1006 / angka2_1006
    print("\nOperator Pembagian")
    print("Hasil =",hasil_1006)

    hasil_1006 = angka1_1006 // angka2_1006
    print("\nOperator Pembagian Bulat")
    print("Hasil =",hasil_1006)

    hasil_1006 = angka1_1006 % angka2_1006
    print("\nOperator Sisa Bagi")
    print("Hasil =",hasil_1006)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_1006 = angka1_1006 ** angka2_1006
print("\nOperator Pangkat")
print("Hasil =",hasil_1006)
````

````python
# Program operator assigment dalam Python

angka1_1006 = int(input("Input angka-1: "))
angka2_1006 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =",angka1_1006)
print("Nilai angka2 =",angka2_1006)

# Assigment biasa
hasil_1006 = angka1_1006 
print("\nAssigment biasa (=)")
print("Hasil =",hasil_1006)

# Assigment penambahan
hasil_1006 = angka1_1006 
hasil_1006 += angka2_1006 
print("\nAssigment penambahan (+=)")
print("Hasil =",hasil_1006)

# Assigment pengurangan
hasil_1006 = angka1_1006 
hasil_1006 -= angka2_1006 
print("\nAssigment pengurangan (-=)")
print("Hasil =",hasil_1006)

# Assigment perkalian
hasil_1006 = angka1_1006 
hasil_1006 *= angka2_1006
print("\nAssigment perkalian (*=)")
print("Hasil =",hasil_1006)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_1006 != 0:
    hasil_1006 = angka1_1006 
    hasil_1006 /= angka2_1006
    print("\nAssigment pembagian (/=)")
    print("Hasil =",hasil_1006)
    # Operator tambahan
    hasil_1006 = angka1_1006 
    hasil_1006 //= angka2_1006
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =",hasil_1006)
    hasil_1006 = angka1_1006 
    hasil_1006 %= angka2_1006
    print("\nOperator sisa bagi (%=)")
    print("Hasil =",hasil_1006)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assigment perpangkatan
hasil_1006 = angka1_1006 
hasil_1006 **= angka2_1006
print("\nOperator perpangkatan (**=)")
print("Hasil =",hasil_1006)
````

### 2.2.3 Program perbandingan_2611531006
</div>

````python
# Program operator perbandingan dalam Python

angka1_1006 = int(input("Input angka-1: "))
angka2_1006 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_1006 = angka1_1006 > angka2_1006
print("\nOperator lebih besar dari")
print("angka1 > angka2 =",hasil_1006)

# Lebih kecil dari
hasil_1006 = angka1_1006 < angka2_1006
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =",hasil_1006)

# Lebih besar dari atau sama dengan
hasil_1006 = angka1_1006 >= angka2_1006
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =",hasil_1006)

# Lebih kecil dari atau sama dengan
hasil_1006 = angka1_1006 <= angka2_1006
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =",hasil_1006)

# Sama dengan
hasil_1006 = angka1_1006 == angka2_1006
print("\nOperator sama dengan")
print("angka1 == angka2 =",hasil_1006)

# Tidak sama dengan
hasil_1006 = angka1_1006 != angka2_1006
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =",hasil_1006)

# Tambahan: perbandingan berantai dalam Python
hasil_1006 = 0 < angka1_1006 < 100
print("\nPerbandingan berantai")
print("0 < angka1_1006 < 100 =",hasil_1006)

hasil_1006 = 0 < angka2_1006 < 100
print("0 < angka2_1006 < 100 =",hasil_1006)
````

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Input Data
```python
angka1_1006 = int(input("Input angka-1: "))
angka2_1006 = int(input("Input angka-2: "))
```
Meminta dua angka dari pengguna. Hasil `input()` berupa string, jadi dibungkus `int()` supaya bisa dibandingkan secara numerik.

2. Enam Operator Perbandingan Dasar
Setiap blok mengikuti pola yang sama: hitung hasil perbandingan, lalu cetak dengan judul dan hasilnya.

| Operator | Arti | Kode |
|---|---|---|
| `>` | lebih besar dari | `angka1_1006 > angka2_1006` |
| `<` | lebih kecil dari | `angka1_1006 < angka2_1006` |
| `>=` | lebih besar atau sama dengan | `angka1_1006 >= angka2_1006` |
| `<=` | lebih kecil atau sama dengan | `angka1_1006 <= angka2_1006` |
| `==` | sama dengan (bukan assignment!) | `angka1_1006 == angka2_1006` |
| `!=` | tidak sama dengan | `angka1_1006 != angka2_1006` |

Variabel `hasil_1006` dipakai berulang-ulang (di-*overwrite* setiap blok), lalu langsung dicetak sebelum pindah ke operator berikutnya.

Catatan kecil: `print("\nOperator lebih besar dari")` — `\n` di awal string membuat baris kosong sebelum judul, supaya outputnya lebih rapi dan terpisah antar blok.

3. Perbandingan Berantai (Chained Comparison)
```python
hasil_1006 = 0 < angka1_1006 < 100
hasil_1006 = 0 < angka2_1006 < 100
```
Fitur khas Python ini setara dengan `(0 < x) and (x < 100)` — mengecek apakah angka berada di rentang 1–99. Di sini dilakukan dua kali: sekali untuk `angka1_1006`, sekali untuk `angka2_1006`.

````python
# Program operator logika dalam Python
# Memasukan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_1006 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_1006 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 = ",a1_1006)
print("A2 = ",a2_1006)

# Konjungsi: bernilai True jika keduanya True
hasil_1006 = a1_1006 and a2_1006
print("\nKonjungsi (AND)")
print("A1 and A2 =",hasil_1006)

# Disjungsi: bernilai True jika salah satunya False
hasil_1006 = a1_1006 or a2_1006
print("\nDisjungsi (OR)")
print("A1 or A2 =",hasil_1006)

# Negasi A1: membalik nilai A1
hasil_1006 = not a1_1006
print("\nNegasi A1(NOT)")
print("not A1",hasil_1006)

# Negasi A2: membalik nilai A2
hasil_1006 = not a2_1006
print("\nNegasi A2(NOT)")
print("not A2",hasil_1006)

# XOR: bernilai True jika kedua nilai berbeda
hasil_1006 = a1_1006 != a2_1006
print("\nDisjungsi Ekslusif (XOR)")
print("A1 XOR A2",hasil_1006)
````

````python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_1006 = int(input("Masukkan angka bitwise-1: "))
angka2_1006 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_1006,"| biner =",bin(angka1_1006))
print("angka1 =",angka2_1006,"| biner =",bin(angka2_1006))

# Bitwise AND
hasil_1006 = angka1_1006 & angka2_1006
print("\nBitwise AND (&)")
print(angka1_1006,"&",angka2_1006,hasil_1006)
print("Biner hasil =",bin(hasil_1006))
print("Biner hasil (8 bit) =",format(hasil_1006,"08b"))

# Bitwise OR
hasil_1006 = angka1_1006 | angka2_1006
print("\nBitwise OR (|)")
print(angka1_1006,"|",angka2_1006,hasil_1006)
print("Biner hasil =",bin(hasil_1006))
print("Biner hasil (8 bit) =",format(hasil_1006,"08b"))

# Bitwise XOR
hasil_1006 = angka1_1006 ^ angka2_1006
print("\nBitwise XOR (^)")
print(angka1_1006,"^",angka2_1006,hasil_1006)
print("Biner hasil =",bin(hasil_1006))
print("Biner hasil (8 bit) =",format(hasil_1006,"08b"))

# Bitwise NOT
hasil_1006 = ~angka1_1006
print("\nBitwise NOT (~)")
print(angka1_1006,"~",angka2_1006,hasil_1006)
print("Biner hasil =",bin(hasil_1006))
print("Biner hasil (8 bit) =",format(hasil_1006,"08b"))

# Bitwise geser kiri
jumlah_geser_1006 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_1006 = angka1_1006 << jumlah_geser_1006
print("\nBitwise geser kiri (<<)")
print(angka1_1006,"<<",jumlah_geser_1006,"=",hasil_1006)
print("Biner hasil =",bin(hasil_1006))
print("Biner hasil (8 bit) =",format(hasil_1006,"08b"))

# Bitwise geser kanan
hasil_1006 = angka1_1006 >> jumlah_geser_1006
print("\nBitwise geser kanan (>>)")
print(angka1_1006,">>",jumlah_geser_1006,"=",hasil_1006)
print("Biner hasil =",bin(hasil_1006))
print("Biner hasil (8 bit) =",format(hasil_1006,"08b"))
````

| Bit A | Bit B | A ^ B |

Operator `>>` menggeser bit ke kanan sebanyak `jumlah_geser_1006` posisi. Untuk bilangan **positif**, ini setara dengan pembagian bulat (*floor division*) dengan 2 pangkat jumlah geseran (`angka // 2ⁿ`).

Operator `&` digunakan untuk **memeriksa bit tertentu**.

Untuk member:

```text
1111
0001
----
0001
```

Hasilnya `0001`, berarti pelanggan adalah member.

Untuk promo:

```text
1111
1000
----
1000
```

Hasilnya `1000`, berarti kode promo tersedia.

### 12. Operator XOR

```python
hasil_xor_1006 = kode_1006 ^ 11
```

Operator `^` digunakan untuk membandingkan perbedaan bit antara kode transaksi dan kode referensi.

```text
1111
1011
----
0100
```

Hasilnya:

```text
0100
```

atau **4 dalam desimal**.

<h1 align="center">BAB III<br>KESIMPULAN</h1>

Berdasarkan praktikum yang telah dilakukan, operator `in` dan `not in` digunakan untuk mengecek keberadaan suatu nilai dalam sebuah data. Operator `in` menghasilkan `True` jika nilai ditemukan, sedangkan `not in` menghasilkan `True` jika nilai tidak ditemukan. Selain itu, operator `is` dan `is not` digunakan untuk membandingkan identitas suatu objek. Pada program, `object1` dan `object2` memiliki identitas yang sama karena merujuk pada objek yang sama.

Praktikum ini juga menunjukkan perbedaan antara `is` dan `==`. Operator `is` membandingkan identitas objek, sedangkan `==` membandingkan isi atau nilai objek. Meskipun `object1` dan `object3` memiliki isi yang sama, keduanya merupakan objek yang berbeda karena `object3` dibuat menggunakan `copy()`. Dengan demikian, pemahaman terhadap operator keanggotaan dan identitas sangat penting agar penggunaan operator dalam Python dapat dilakukan dengan tepat sesuai kebutuhan.
