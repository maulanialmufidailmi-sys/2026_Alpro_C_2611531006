## 2.1 Dasar Teori

**Perulangan (Looping)** adalah struktur kontrol yang menjalankan blok instruksi secara berulang selama kondisi tertentu terpenuhi. Python memiliki `for` untuk iterasi dengan jumlah yang sudah diketahui dan `while` untuk iterasi yang bergantung pada suatu kondisi.

**Nested Loop** adalah perulangan yang berada di dalam perulangan lain, di mana perulangan dalam dijalankan sepenuhnya pada setiap iterasi perulangan luar.

**Perhitungan Jumlah Bilangan Genap** adalah proses menjumlahkan bilangan bulat yang habis dibagi dua (sisa bagi sama dengan nol, dicek dengan operator `%`) dengan memanfaatkan perulangan, percabangan `if`, dan variabel akumulator.
## 2.2 Kode Program

### 2.2.1 Program_Perulangan_for1_2611531006
</div>

```python
# Buat file dengan nama Perulangan_for1_2611531006.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()


ulang_1006 = int(input("Masukkan jumlah perulangan: "))
for i_1006 in range(ulang_1006):
    print(f"Perulangan ke-{i_1006}")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

Program ini meminta pengguna memasukkan sebuah angka, lalu mencetak tulisan "Perulangan ke-..." sebanyak angka tersebut menggunakan perulangan `for`.

**Baris per baris**

1. ulang_1006 = int(input("Masukkan jumlah perulangan: "))
   - input() menampilkan pesan dan membaca masukan dari pengguna. Hasilnya selalu berupa string.
   - int() mengubah string tersebut menjadi bilangan bulat agar bisa dipakai pada range().
   - Hasilnya disimpan di variabel ulang_1006. Akhiran 1006 adalah 4 digit terakhir NIM, sesuai ketentuan penamaan variabel.

2. for i_1006 in range(ulang_1006):
   - range(ulang_1006) menghasilkan deretan bilangan dari 0 sampai ulang_1006 - 1.
   - Variabel i_1006 menampung nilai satu per satu di setiap putaran, sehingga perulangan berjalan sebanyak ulang_1006 kali.

3. print(f"Perulangan ke-{i_1006}")
   - Baris ini berada di dalam blok `for` (ditandai indentasi), sehingga dijalankan di setiap putaran.
   - Huruf f di depan tanda kutip menandakan *f-string*, yaitu string yang memungkinkan nilai variabel disisipkan lewat kurung kurawal {}.

### 2.2.3 Perulangan_for3_2611531006
</div>

```python
# Buat nama Perulangan_for3_2611531006.py

ulang_1006 = int(input("Masukkan jumlah perulangan: "))

jumlah_1006 = 0
for i_1006 in range(1, ulang_1006 + 1):
    print(i_1006, end=" ")
    jumlah_1006 = jumlah_1006 + 1

    if 1 < ulang_1006:
        print(" + ", end=" ")
    else:
        print(" + ", jumlah_1006, end=" ")
print()
print("Jumlah =", jumlah_1006)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

1. Fungsi int(input(...)) membaca masukan sebagai bilangan bulat dan menyimpannya di ulang_1006.
2. Variabel jumlah_1006 diberi nilai awal 0 sebagai penampung.
3. Perulangan for dengan range(1, ulang_1006 + 1) berjalan dari 1 sampai ulang_1006. Di setiap putaran, angka dicetak sejajar dalam satu baris (end=" "), lalu jumlah_1006 bertambah 1.
4. Percabangan if mencetak tanda plus setelah angka. Jika kondisinya tidak terpenuhi, cabang else mencetak tanda plus beserta nilai jumlah_1006.
5. print() kosong pindah ke baris baru, lalu hasil akhir ditampilkan dengan print("Jumlah =", jumlah_1006).

Jika pengguna memasukkan 3, keluarannya kira-kira: 1 + 2 + 3 + lalu pada baris berikutnya Jumlah = 3.

Ada dua hal yang perlu diperhatikan:
- Kondisi if 1 < ulang_1006 membandingkan angka 1 dengan jumlah perulangan, bukan dengan i_1006. Akibatnya, jika n lebih dari 1, tanda plus selalu tercetak, termasuk setelah angka terakhir. Agar tanda plus tidak muncul di angka terakhir, kondisinya sebaiknya if i_1006 < ulang_1006.
- Variabel jumlah_1006 hanya bertambah 1 di setiap putaran, sehingga isinya adalah banyaknya perulangan (3 untuk n = 3), bukan hasil penjumlahan 1 + 2 + 3 = 6. Jika yang diinginkan hasil penjumlahan, barisnya diubah menjadi jumlah_1006 = jumlah_1006 + i_1006.

### 2.2.4 Nested_for1_2611531006
</div>

```python
# Buat file dengan nama nested_for1_2611531006.py

batas_1006 = int(input("Masukkan nilai batas: "))
for line_1006 in range(1, batas_1006 + 1):
    for j_1006 in range(1, (-1 * line_1006 + batas_1006) + 1):
        print(".", end=" ")
    print(line_1006,)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

2. **Perulangan luar:** for line_1006 in range(1, batas_1006 + 1) berjalan dari 1 sampai batas_1006. Variabel line_1006 menyimpan nomor baris yang sedang diproses.
3. **Perulangan dalam:** for j_1006 in range(1, (-1 * line_1006 + batas_1006) + 1) berjalan sebanyak batas_1006 dikurangi line_1006 kali. Karena itu, makin besar nomor baris, makin sedikit titik yang dicetak.
4. print(".", end=" ") mencetak titik tanpa pindah baris, sehingga titik-titik tercetak berdampingan.
5. print(line_1006,) berada di luar perulangan dalam, jadi nomor baris dicetak sekali setelah semua titik selesai, lalu pindah ke baris baru.

### 2.2.7 Nested_for4_2611531006

```python
# Buat file dengan nama nested_for4_2611531006.py

tinggi_1006 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1006 % 2 !=0:
    print("Tinggi harus bilangan genap!")
else:
    a_1006 = tinggi_1006
    c_1006 = a_1006
    lebar_1006 = (2 * tinggi_1006) - 2
   
    for i_1006 in range(1, tinggi_1006 + 1):
        b_1006 = c_1006 + 1

        for j_1006 in range(1, lebar_1006 + 1):

            #Baris atas dan bawah
            if i_1006 == 1 or i_1006 == tinggi_1006:
                if j_1006 == 1 or j_1006 == lebar_1006:
                    print("#", end="")
                else:
                    print("=", end="")

            #Baris isi
            else:
                if j_1006 == 1 or  j_1006 == lebar_1006:
                    print("|", end="")
                else:
                    if j_1006 == c_1006:
                        print("<", end="")
                    elif j_1006 == b_1006:
                        print(">", end="")
                    elif j_1006 == (lebar_1006 - c_1006):
                        print("<", end="")
                    elif j_1006 == (lebar_1006 - c_1006 + 1):
                        print(">", end="")
                    elif j_1006 > b_1006 and j_1006< (lebar_1006 - c_1006):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_1006 -= 2

        if a_1006 <= 0:
            c_1006 = (-a_1006) +2
        else:
            c_1006 = a_1006
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

1. **Tujuan program.** Program mencetak pola persegi panjang berbingkai. Di dalamnya ada bentuk belah ketupat yang tersusun dari tanda < dan >, dengan titik mengisi bagian tengahnya.
2. **Masukan pengguna.** Fungsi input() membaca ketikan pengguna sebagai string, lalu int() mengubahnya menjadi bilangan bulat. Hasilnya disimpan di tinggi_1006 dan menentukan banyaknya baris pola.
3. **Validasi bilangan genap.** Kondisi tinggi_1006 % 2 != 0 memeriksa sisa bagi dua. Jika sisanya bukan 0, tinggi itu ganjil, sehingga program hanya mencetak peringatan dan tidak membuat pola. Pola hanya simetris jika tingginya genap.
4. **Variabel awal.** a_1006 dan c_1006 diisi sama dengan tinggi. Lebar dihitung dengan rumus 2 × tinggi dikurangi 2, jadi untuk tinggi 10 lebarnya 18 kolom.
5. **Nested loop.** Perulangan luar (i_1006) berjalan sekali untuk setiap baris, sedangkan perulangan dalam (j_1006) berjalan penuh untuk mencetak seluruh kolom pada baris itu. print() kosong di akhir pindah ke baris baru.
6. **Parameter end="".** Biasanya print pindah baris setelah mencetak. Dengan end diisi string kosong, karakter berikutnya tercetak langsung di sebelahnya tanpa spasi, sehingga bentuk pola tidak terputus.
7. **Baris atas dan bawah.** Pada baris pertama dan terakhir, kolom paling kiri dan paling kanan dicetak # sebagai pojok bingkai. Semua kolom di antaranya dicetak = sebagai garis atas dan bawah.
8. **Dinding samping.** Pada baris isi, kolom pertama dan kolom terakhir dicetak | sebagai dinding kiri dan kanan bingkai.
9. **Tanda belah ketupat.** Di sisi kiri, kolom c_1006 dicetak < dan kolom setelahnya (b_1006) dicetak >. Di sisi kanan, pasangan serupa dicetak pada kolom lebar dikurangi c_1006 dan kolom setelahnya. Dua pasangan inilah yang membentuk garis tepi belah ketupat.
10. **Isi dan ruang kosong.** Kolom di antara pasangan kiri dan pasangan kanan dicetak titik sebagai isi bentuk. Kolom di luar pasangan, yaitu antara dinding dan tanda, dicetak spasi.
11. **Pergeseran pola.** Di akhir setiap baris, a_1006 dikurangi 2. Selama a_1006 masih positif, c_1006 sama dengan a_1006, sehingga nilainya turun dan tanda bergerak dari tengah menuju tepi. Setelah a_1006 mencapai nol atau negatif, c_1006 menjadi minus a_1006 ditambah 2, sehingga nilainya naik dan tanda kembali ke tengah. Gerak keluar lalu kembali inilah yang membentuk belah ketupat.
12. **Contoh tinggi 10.** Nilai c_1006 pada baris ke-2 sampai ke-9 berturut-turut adalah 8, 6, 4, 2, 2, 4, 6, 8. Baris ke-5 dan ke-6 sama dan menjadi bagian terlebar, karena tanda berada paling dekat dengan dinding. Baris ke-2 dan ke-9 tidak memiliki titik karena kedua pasangan tanda berdempetan di tengah. Baris ke-1 dan ke-10 adalah bingkai atas dan bawah.


### 2.2.8 Jumlah_genap_2611531006

```python
# Buat file dengan nama jumlah_genap_2611531006.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_1006 = int(input("Maukkan nilai batas: "))

jumlah_1006 = 0
for i_1006 in range(1, ulang_1006 + 1):
    if i_1006 % 2 == 0:
        print(i_1006, end="")
        jumlah_1006 = jumlah_1006 + i_1006
        
        if i_1006 < ulang_1006:
            print("+", end="")
        else:
            print("=", jumlah_1006, end="")
print()
print("Jumlah =", jumlah_1006)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

**Penjelasan :** 
1. ulang_1006 = int(input("Maukkan nilai batas: "))
Meminta input angka dari pengguna melalui `input()`, lalu mengonversinya menjadi bilangan bulat (*integer*) dengan int() dan menyimpannya di variabel `ulang_1006`.
2. jumlah_1006 = 0
Menyiapkan variabel `jumlah_1006` dengan nilai awal `0` yang berfungsi sebagai penampung (*accumulator*) total penjumlahan.
3. for i_1006 in range(1, ulang_1006 + 1):
Melakukan perulangan (*loop*) dengan variabel `i_1006` mulai dari angka `1` hingga `ulang_1006`.
4. if i_1006 % 2 == 0:
Memeriksa apakah angka `i_1006` saat ini adalah bilangan genap dengan mengecek sisa bagi (`% 2`) bernilai `0`.
5. print(i_1006, end="")
Mencetak angka genap yang ditemukan tanpa membuat baris baru karena adanya atribut `end=""`.
6. jumlah_1006 = jumlah_1006 + i_1006
Menambahkan nilai angka genap yang sedang diproses ke dalam total akumulasi pada variabel `jumlah_1006`.
7. if i_1006 < ulang_1006:
Memeriksa apakah posisi angka genap saat ini masih kurang dari angka batas akhir.
8. print("+", end="")
Mencetak karakter `+` tanpa ganti baris jika angka genap belum mencapai batas akhir.
9. else:
Menentukan bagian tindakan alternatif jika perulangan angka genap telah mencapai batas akhir.
10. print("=", jumlah_1006, end="")
Mencetak karakter `=` beserta nilai total `jumlah_1006` saat perulangan berada di angka batas akhir.
11. print()
Mencetak baris baru kosong untuk merapikan tampilan output.
12. print("Jumlah =", jumlah_1006)
Menampilkan teks hasil akhir total penjumlahan seluruh bilangan genap ke layar.

```python
# Nama Berkas: tugas5_2611531006.py
# Program Pola Jam Pasir Kristal Palindromik Berbingkai

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

# 1. Input nilai N secara dinamis
n_1006 = int(input("Masukkan ukuran skala jam pasir (N): "))

# 2. Bingkai Pembatas Horizontal Atas
print("#", end="")
for i_1006 in range(4 * n_1006 + 5):
    print("=", end="")
print("#")

# 3. Fase 1: Jam Pasir Atas (Baris N turun sampai 1)
for baris_1006 in range(n_1006, 0, -1):
    print("| ", end="")
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_1006 in range(2 * (n_1006 - baris_1006)):
        print(" ", end="")
        
    # Deret angka mundur (baris turun ke 1)
    for angka_1006 in range(baris_1006, 0, -1):
        print(angka_1006, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju (1 naik ke baris)
    for angka_1006 in range(1, baris_1006 + 1):
        print(" ", end="")
        print(angka_1006, end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_1006 in range(2 * (n_1006 - baris_1006)):
        print(" ", end="")
        
    print(" |")

# 4. Fase 2: Poros Titik Pusat Jam Pasir (Singularity)
print("|", end="")

# Spasi penyeimbang kiri: 2 * N + 1
for spasi_1006 in range(2 * n_1006 + 1):
    print(" ", end="")
    
print("<*>", end="")

# Spasi penyeimbang kanan: 2 * N + 1
for spasi_1006 in range(2 * n_1006 + 1):
    print(" ", end="")
    
print("|")

# 5. Fase 3: Jam Pasir Bawah (Baris 1 naik sampai N)
for baris_1006 in range(1, n_1006 + 1):
    print("| ", end="")
    
    # Spasi penyeimbang kiri: 2 * (N - baris)
    for spasi_1006 in range(2 * (n_1006 - baris_1006)):
        print(" ", end="")
        
    # Deret angka mundur (baris turun ke 1)
    for angka_1006 in range(baris_1006, 0, -1):
        print(angka_1006, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju (1 naik ke baris)
    for angka_1006 in range(1, baris_1006 + 1):
        print(" ", end="")
        print(angka_1006, end="")
        
    # Spasi penyeimbang kanan: 2 * (N - baris)
    for spasi_1006 in range(2 * (n_1006 - baris_1006)):
        print(" ", end="")
        
    print(" |")

# 6. Bingkai Pembatas Horizontal Bawah
print("#", end="")
for i_1006 in range(4 * n_1006 + 5):
    print("=", end="")
print("#")
```

Penjelasan:

1. **Pembacaan Input**
* Program menerima ukuran skala $N$ via `input()`, dikonversi ke *integer* `int()`, dan disimpan di variabel `n_1006`.


2. **Bingkai Pembatas Horizontal Atas**
* Karakter `#` dicetak di ujung kiri tanpa ganti baris (`end=""`).
* Perulangan `for` mencetak karakter `=` sebanyak $4N + 5$ kali.
* Karakter `#` dicetak di ujung kanan diikuti ganti baris.


3. **Fase 1: Jam Pasir Atas (Perulangan Luar `n_1006` turun ke `1`)**
* Garis Pembatas Kiri: Mencetak pembatas `| `.
* Spasi Penyeimbang Kiri: Perulangan `for` mencetak karakter spasi sebanyak $2 \times (N - \text{baris})$ kali.
* Deret Angka Mundur: Perulangan `for` mencetak angka dari nilai `baris` turun ke `1`, dipisahkan spasi.
* Poros Kristal: Mencetak karakter `<*>`.
* Deret Angka Maju: Perulangan `for` mencetak angka dari `1` naik ke nilai `baris`, diawali spasi.
* Spasi Penyeimbang Kanan: Perulangan `for` mencetak karakter spasi sebanyak $2 \times (N - \text{baris})$ kali.
* Garis Pembatas Kanan: Mencetak pembatas ` |` lalu ganti baris.


4. **Fase 2: Poros Titik Pusat / Singularity**
* Garis Pembatas Kiri: Mencetak pembatas `|`.
* Spasi Penyeimbang Kiri: Perulangan `for` mencetak karakter spasi sebanyak $2N + 1$ kali.
* Poros Kristal: Mencetak karakter `<*>`.
* Spasi Penyeimbang Kanan: Perulangan `for` mencetak karakter spasi sebanyak $2N + 1$ kali.
* Garis Pembatas Kanan: Mencetak pembatas `|` lalu ganti baris.


5. **Fase 3: Jam Pasir Bawah (Perulangan Luar `1` naik ke `n_1006`)**
* Melakukan eksekusi perulangan internal yang sama persis dengan **Fase 1**, tetapi urutan perulangan luar berjalan naik dari `1` hingga `N` untuk membentuk pencerminan ke bawah.


6. **Bingkai Pembatas Horizontal Bawah**
* Karakter `#` dicetak di ujung kiri.
* Perulangan `for` mencetak karakter `=` sebanyak $4N + 5$ kali.
* Karakter `#` dicetak di ujung kanan untuk menutup bingkai.
